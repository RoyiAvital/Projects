## Generate Data


### Extract Dividends from Yahoo Finance

1. Open the Yahoo Finance website and navigate to the stock's page (See [Yahoo Finance - Coca Cola](https://finance.yahoo.com/quote/KO)).
2. Click on the "Historical Data" tab.
3. Choose `Dividends Only` from the "Show" dropdown menu.
4. Set the time period to `Max` to get all available dividend data.
5. Click on the table data and look for the `Inspect` to see the HTML structure of the data.
6. Find the `table` element that contains the dividend data and copy its HTML content.
7. Right click on the element and choose `Copy` > `Outer HTML` to copy the entire table's HTML.
8. Paste the copied HTML into a text file and save it as `<FileName>.txt`.
9. Use `ParseYahooFinanceDivTable` to parse it into a Data Frame.

````python
# Load / Generate Data

dfDiv = ParseYahooFinanceDivTable(yahooDivTable)
# Filter the DataFrame to include only rows with 'Date' >= startDate
dfDiv = dfDiv[dfDiv['Date'] >= startDate]
````

````python
dfStock = ParseYahooFinanceCloseTable('CloseYahoo_KO_2026_04_18.txt')
# Filter the DataFrame to include only rows with 'Date' >= startDate
dfStock = dfStock[dfStock['Date'] >= startDate]
````

````python
dsSwing = (dfStock['High'] - dfStock['Low']) / dfStock['Open']
````

````python
# Create a DF with the Dividends and the Close Price of that particular day
dfDivClose = dfDiv.merge(dfStock[['Date', 'AdjClose']], on = 'Date', how = 'left')
````

````python
dfDivClose
````

````python
dsDivRel = dfDivClose['Dividend'] / dfDivClose['AdjClose']
````

````python
hF, hA = plt.subplots(figsize = (6, 4))
sns.histplot(dsDivRel, stat = 'probability', bins = 30, kde = True, ax = hA)
hA.set_title('Dividend Relative to Adjusted Close Price Distribution')
hA.set_xlabel('Dividend / Adjusted Close Price')
hA.axvline(x = dsDivRel.mean(), color = 'r', linestyle = '--', label = f'Mean: {dsDivRel.mean():0.4f}')
hA.axvline(x = dsDivRel.quantile(0.25), color = 'b', linestyle = '--', label = f'Percentile 25%: {dsDivRel.quantile(0.25):0.4f}')
hA.axvline(x = dsDivRel.median(), color = 'g', linestyle = '--', label = f'Median: {dsDivRel.median():0.4f}')
hA.axvline(x = dsDivRel.quantile(0.75), color = 'b', linestyle = '--', label = f'Percentile 75%: {dsDivRel.quantile(0.75):0.4f}')
hA.legend();
````

````python
hF, hA = plt.subplots(figsize = (6, 4))
sns.histplot(dsSwing, stat = 'probability', bins = 30, kde = True, ax = hA)
hA.set_title('Daily Swing Relative to Open Price Distribution')
hA.set_xlabel('Daily Swing / Open Price')
hA.axvline(x = dsSwing.mean(), color = 'r', linestyle = '--', label = f'Mean: {dsSwing.mean():0.4f}')
hA.axvline(x = dsSwing.quantile(0.25), color = 'b', linestyle = '--', label = f'Percentile 25%: {dsSwing.quantile(0.25):0.4f}')
hA.axvline(x = dsSwing.median(), color = 'g', linestyle = '--', label = f'Median: {dsSwing.median():0.4f}')
hA.axvline(x = dsSwing.quantile(0.75), color = 'b', linestyle = '--', label = f'Percentile 75%: {dsSwing.quantile(0.75):0.4f}')
hA.legend();
````
